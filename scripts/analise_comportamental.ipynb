# Análise Descritiva do Engajamento
Investigação estatística do número de rodadas jogadas para identificar a distribuição da base, assimetria entre média e mediana, e presença de outliers.

```python
mediana_rodada = df["sum_gamerounds"].median()
print(f"Meidana de rodadas: {mediana_rodada}")
```

```python
analise_grupos = df.groupby('version')[['sum_gamerounds', 'retention_1', 'retention_7']].mean()
display(analise_grupos)
```

```python
df["sum_gamerounds"].quantile([0.90, 0.95, 0.99, 0.999])
df.nlargest(10, "sum_gamerounds")
```

```python
df_sem_outlier = df[df["sum_gamerounds"] < 49854]

df_sem_outlier.groupby("version")[["sum_gamerounds", "retention_1", "retention_7"]].mean()
```
